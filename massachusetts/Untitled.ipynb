```python
import pandas as pd

injury_types = [
    "Non-fatal injury - Incapacitating",
    "Fatal injury (K)",
    "Deceased not caused by crash",
]

# 2016
data16 = pd.read_csv("mass2016.csv")
severe16 = data16[data16["Injury Type"].isin(injury_types)]["Crash Number"]
data16 = data16[data16["Crash Number"].isin(severe16)]

# 2017
data17 = pd.read_csv("mass2017.csv")
severe17 = data17[data17["Injury Type"].isin(injury_types)]["Crash Number"]
data17 = data17[data17["Crash Number"].isin(severe17)]

# 2018
data18 = pd.read_csv("mass2018.csv")
severe18 = data18[data18["Injury Type"].isin(injury_types)]["Crash Number"]
data18 = data18[data18["Crash Number"].isin(severe18)]

# 2019
data19 = pd.read_csv("mass2019.csv")
severe19 = data19[data19["Injury Type"].isin(injury_types)]["Crash Number"]
data19 = data19[data19["Crash Number"].isin(severe19)]

# 2020
data20 = pd.read_csv("mass2020.csv")
severe20 = data20[data20["Injury Type"].isin(injury_types)]["Crash Number"]
data20 = data20[data20["Crash Number"].isin(severe20)]

# 2021
data21 = pd.read_csv("mass2021.csv")
severe21 = data21[data21["Injury Type"].isin(injury_types)]["Crash Number"]
data21 = data21[data21["Crash Number"].isin(severe21)]

# 2022
data22 = pd.read_csv("mass2022.csv")
severe22 = data22[data22["Injury Type"].isin(injury_types)]["Crash Number"]
data22 = data22[data22["Crash Number"].isin(severe22)]

# 2023
data23 = pd.read_csv("mass2023.csv")
severe23 = data23[data23["Injury Type"].isin(injury_types)]["Crash Number"]
data23 = data23[data23["Crash Number"].isin(severe23)]

# 2024
data24 = pd.read_csv("mass2024.csv")
severe24 = data24[data24["Injury Type"].isin(injury_types)]["Crash Number"]
data24 = data24[data24["Crash Number"].isin(severe24)]

# 2025
data25 = pd.read_csv("mass2025.csv")
severe25 = data25[data25["Injury Type"].isin(injury_types)]["Crash Number"]
data25 = data25[data25["Crash Number"].isin(severe25)]

# 2026
data26 = pd.read_csv("mass2026.csv")
severe26 = data26[data26["Injury Type"].isin(injury_types)]["Crash Number"]
data26 = data26[data26["Crash Number"].isin(severe26)]

# Combine all 11 years into a single DataFrame
mass = pd.concat(
    [
        data16,
        data17,
        data18,
        data19,
        data20,
        data21,
        data22,
        data23,
        data24,
        data25,
        data26,
    ],
    axis=0,
    ignore_index=True,
)

# Export combined output
mass.to_csv("mass.csv", index=False)
```
