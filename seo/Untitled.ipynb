```python
import pandas as pd
```

```python
data = pd.read_csv("us_cities_states_counties_zips.csv", delimiter="|")
```

```python
arizona = data[data['State short']=="AZ"]
cities = arizona['City'].unique()
for city in cities:
    print(f"<p style='display:none'>{city} Accident Data, {city} Vision Zero Map</p>")

counties = arizona['County'].unique()
for county in counties:
    print(f"<p style='display:none'>{county} COUNTY Accident Data, {county} COUNTY Vision Zero Map map</p>")

aliases = arizona['City alias'].unique()
for alias in aliases:
    if alias not in cities:
        print(f"<p style='display:none'>{alias} Accident Data, {alias} Vision Zero Map</p>")

zips = set()
for x in arizona.index:
    zippies = arizona['City zip codes'][x].split(" ")
    for zip in zippies:
        zips.add(zip)
for zcode in zips:
    print(f"<p style='display:none'>{zcode} Accident Data, {zcode} Vision Zero Map map</p>")
```

```python
# Mapping state names to their official 2-letter postal codes
states_map = {
    "arizona": "AZ",
    "california": "CA",
    "colorado": "CO",
    "florida": "FL",
    "illinois": "IL",
    "michigan": "MI",
    "massachusetts": "MA",
    "new york": "NY",
    "new jersey": "NJ",
    "north carolina": "NC",
    "pennsylvania": "PA",
    "texas": "TX",
    "tennessee": "TN",
    "virginia": "VA",
    "washington": "WA",
}

for state_name, state_code in states_map.items():
    # Filter dataset for the current state
    state_df = data[data["State short"] == state_code]

    # Open a text file for writing (creates state_name.txt)
    with open(f"{state_name}.txt", "w", encoding="utf-8") as f:

        # 1. Cities
        cities = state_df["City"].dropna().unique()
        for city in cities:
            f.write(
                f"<p style='display:none'>{city} Accident Data, {city} Vision Zero Map</p>\n"
            )

        # 2. Counties
        counties = state_df["County"].dropna().unique()
        for county in counties:
            f.write(
                f"<p style='display:none'>{county} COUNTY Accident Data, {county} COUNTY Vision Zero Map map</p>\n"
            )

        # 3. City Aliases (excluding existing cities)
        aliases = state_df["City alias"].dropna().unique()
        for alias in aliases:
            if alias not in cities:
                f.write(
                    f"<p style='display:none'>{alias} Accident Data, {alias} Vision Zero Map</p>\n"
                )

        # 4. ZIP Codes
        zips = set()
        # Drop NaN values before iterating to prevent string split errors
        valid_zips = state_df["City zip codes"].dropna()
        for zippie_str in valid_zips:
            for zcode in str(zippie_str).split():
                zips.add(zcode)

        for zcode in zips:
            f.write(
                f"<p style='display:none'>{zcode} Accident Data, {zcode} Vision Zero Map map</p>\n"
            )
```
